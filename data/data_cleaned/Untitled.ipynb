```python
import pandas as pd

# Load your datasets
cd = pd.read_csv("circRNA_disease_clean.csv")
cm = pd.read_csv("circRNA_miRNA_clean.csv")
md = pd.read_csv("miRNA_disease_clean.csv")

# Extract unique nodes
circRNAs = set(cd['circRNA']).union(set(cm['circRNA']))
miRNAs = set(cm['miRNA']).union(set(md['miRNA']))
diseases = set(cd['disease']).union(set(md['disease']))

print("Unique circRNAs:", len(circRNAs))
print("Unique miRNAs:", len(miRNAs))
print("Unique diseases:", len(diseases))
```
